```python
# 명령어를 순차적으로 받아서 제목, 내용 입력
# 생성 및 저장
# 지속적인 명령어 사용 가능, 그리고 break 기능을 명령어로 입력하여 반복문 종료
# 번호/ 제목/ 내용을 저장한 리스트 출력

# 게시글 삭제 기능 구현
# article delete 1
# 게시글 수정 기능 구현
# article edit 1
# 게시글 상세보기 기능 구현
# article detail 1
import time

running = True
i = 1
#게시글 리스트
Articles = []
#회원 리스트
Members = []
login = False


print("===CLI 게시판 실행===")
while running:
  # 명령어 입력
  # ============================================================================
  command = input("명령어 입력:").strip()

  # exit를 입력하면 종료
  if command == "exit":
      print("===CLI 게시판 종료===")
      break

  #로그인 (아이디, 비밀번호) 입력 요구
  # ============================================================================
  elif command == 'login':
    #logincheck[0] = Username , logincheck[1] = password
    logincheck = input("아이디, 비밀번호를 입력해주세요.").split(",")
    found = False

    # 회원가입된 딕셔너리 중에 일치하는 것이 있는가?
    # 반복문을 이용해, 모두 비교
    for m in Members:

      if logincheck[0] == str(m.get("Username")) and logincheck[1] == str(m.get("Password")):
        print("로그인 되었습니다.")
        login = True
        break

      elif found == False:
        print("회원정보와 다릅니다.")
      continue


  #회원 가입 (아이디, 비밀번호)
  # ============================================================================
  elif command.startswith("sign up"):
    Member = {"Username":input("아이디 입력"), "Password":input("비밀번호 입력")}
    Members.append(Member)




  # article write를 입력하면 제목/내용 게시글 입력
  # ============================================================================
  elif command == "article write":
    # 로그인 되어 있는지 확인
    if login == False:
      print("로그인되어 있지 않아, 접근할 수 없습니다. 회원가입 - Sign up를 입력하세요. ")
      continue


    # 제목 입력
    print("Title write:")
    command_title = input()

    # 내용 입력
    print("Content write:")
    command_content = input()

  # 제목/ 내용 저장한 input()값 출력
    print(f"제목: ({command_title})")
    print(f"내용: ({command_content})")
    print(f"{i}번 글이 생성되었습니다.")

  #딕셔너리에 분류 (번호, 제목, 내용, 시간)
    Article = {"num":i, "title":command_title, "content":command_content, "time":time.strftime('%Y-%m-%d %H:%M:%S')}
    Articles.append(Article) # 리스트로 저장

    i += 1

  # 게시글 리스트 기능
  # ============================================================================
  elif command == "article list":
    # 로그인 되어 있는지 확인
    if login == False:
      print("로그인되어 있지 않아, 접근할 수 없습니다. 회원가입 - Sign up를 입력하세요. ")
      continue


    if not Articles:
      print("게시글이 없다.")
      continue

    Articles.reverse()  #내용 리버스하기 - 최신순이 위로 올라가게
    print("=======================")
    print("번호   /   제목   /   내용  /  작성일")
    for j in Articles:
      print(f"{j.get('num')} / {j.get('title')} / {j.get('content')} / {j.get('time')}")
    print("=======================")


  # 게시글 삭제 기능
  # ============================================================================
  elif command.startswith("article delete"):
    commandBits = command.split(" ")
    deletedid = int(commandBits[2])
    found = False

    # 로그인 되어 있는지 확인
    if login == False:
      print("로그인되어 있지 않아, 접근할 수 없습니다. 회원가입 - Sign up를 입력하세요. ")
      continue

    # 기입해야할 명령어 수가 부족한 경우
    if len(command) < 3:
      print("명령어의 수가 부족합니다.")
      continue

    #만약 num을 써야할 곳에 문자열을 기입 했을 경우
    if not commandBits[2].isdigit():
      print("해당 게시글 num이 존재하지 않습니다.")
      continue

    # d에 articeles 리스트를 하나씩 넣는다.
    for d in Articles:
      if d.get('num') == deletedid:
        Articles.remove(d)
        print(f"{commandBits[2]}을 지웠습니다.")
        found = True
      break

    #만약 num이 존재 하지 않는 num인 경우
    if found == False:
      print(f"{commandBits[2]}번 글은 존재하지 않습니다.")
      continue


  #게시글 수정 기능
  # ============================================================================
  elif command.startswith("article edit"):
    commandBits = command.split(" ")
    Editnumber = int(commandBits[2])
    Editcommand = (input("수정하는 것이 title인지 content인지 입력하세요.:"))
    found = False

    # 로그인 되어 있는지 확인
    if login == False:
      print("로그인되어 있지 않아, 접근할 수 없습니다. \n[로그인]을 해주세요. [회원가입] - Sign up를 입력하세요. ")
      continue

    # 기입해야할 명령어 수가 부족한 경우
    if len(command) < 3:
      print("명령어의 수가 부족합니다.")
      continue

    #만약 num을 써야할 곳에 문자열을 기입 했을 경우
    if not commandBits[2].isdigit():
      print("해당 게시글 num이 존재하지 않습니다.")
      continue

    #만약 num이 존재 하지 않는 num인 경우
    if int(commandBits[2]) != len(Articles):
        print(f"{commandBits[2]}번 글은 존재하지 않습니다.")
        continue

    for e in Articles:
      # 번호 입력 key와 수정하고 싶은 번호가 같은지 확인
      if e.get('num') == Editnumber:
        #리스트 내부에 있는 인덱스 값의 딕셔너리의 key 값을 수정
        Articles[Editnumber-1][f"{Editcommand}"] = input("수정할 내용 입력하세요.:")
        print(f"{Editnumber}번의 글에서 {Editcommand}를 수정했습니다.")
        found = True

    if found == False:
      print(f"{Editnumber}번 글은 존재하지 않습니다.")
      continue

  #게시글 상세보기 (번호, 제목, 내용, 날짜 )
  # ============================================================================
  elif command.startswith("article detail"):
    commandBits = command.split(" ")
    Detailnumber = int(commandBits[2])
    found = False

    # 로그인 되어 있는지 확인
    if login == False:
      print("로그인되어 있지 않아, 접근할 수 없습니다. 회원가입 - Sign up를 입력하세요. ")
      continue

    # 기입해야할 명령어 수가 부족한 경우
    if len(command) < 3:
      print("명령어의 수가 부족합니다.")
      continue

    #만약 num을 써야할 곳에 문자열을 기입 했을 경우
    if not commandBits[2].isdigit():
      print("해당 게시글 num이 존재하지 않습니다.")
      continue

    for de in Articles:
      if de.get('num') == Detailnumber:
        print(f"번호: {de.get('num')} \n제목: {de.get('title')} \n내용: {de.get('content')} \n시간: {de.get('time')}")
        found = True

    if found == False:
      print(f"{Detailnumber}번 글은 존재하지 않습니다.")
      continue

  # 일치하지 않는 명령어일 경우 출력
  else:
    print("지원하지 않는 명령어입니다.")
```
