```python
# 명령어를 순차적으로 받아서 제목, 내용 입력
# 생성 및 저장
# 지속적인 명령어 사용 가능, 그리고 break 기능을 명령어로 입력하여 반복문 종료
# 번호/ 제목/ 내용을 저장한 리스트 출력
running = True
i = 1
write_list = []
write_list.append("=======================")
write_list.append("번호   /   제목   /   내용")
New_list=[]
space = "  "


print("===CLI 게시판 실행===")
while running:
  # 번호 갱신
  command = input("명령어 입력:")

  # exit를 입력하면 종료
  if command == "exit":
      print("===CLI 게시판 종료===")
      break

  # article write를 입력하면 제목/내용 게시글 입력
  elif command == "article write":

    # 제목 입력
    print("Title write:")
    command_title = input()

    # 내용 입력
    print("Content write:")
    command_content = input()

  # 제목/ 내용 저장한 input()값 출력
    print(f"제목: ({command_title})")
    print(f"내용: ({command_content})")
    print(f"{i}번 글이 생성되었습니다.")
  #리스트에 저장 (번호/제목/내용)
    New_list.append(f" {space}{i}   /   {command_title}   /   {command_content}")
    i += 1

  # 저장해둔 리스트 값 차례대로 출력
  elif command == "article list":
    #내용 리버스하기 - 최신순이 위로 올라가게
    New_list.reverse()
    # 게시글 저장한 것과 기본 리스트에 더하기
    write_list += New_list
    write_list.append("=======================")
    for j in write_list:
      print(j)

  # 일치하지 않는 명령어일 경우 출력
  else:
    print("지원하지 않는 명령어입니다.")
```
