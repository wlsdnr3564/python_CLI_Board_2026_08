```python
# 명령어를 순차적으로 받아서 제목, 내용 입력
# 생성 및 저장
# 지속적인 명령어 사용 가능, 그리고 break 기능을 명령어로 입력하여 반복문 종료
# 번호/ 제목/ 내용을 저장한 리스트 출력

# 게시글 삭제 기능 구현
# article delete 1
# 게시글 수정 기능 구현
# article edit 1
# 게시글 상세보기 기능 구현
# article detail 1


running = True
i = 1
Articles = []


print("===CLI 게시판 실행===")
while running:
  # 번호 갱신
  command = input("명령어 입력:").split()

  # exit를 입력하면 종료
  if "exit" in command:
      print("===CLI 게시판 종료===")
      break

  # article write를 입력하면 제목/내용 게시글 입력
  # elif command == "article write":
  elif "article_write" in command:

    # 제목 입력
    print("Title write:")
    command_title = input()

    # 내용 입력
    print("Content write:")
    command_content = input()

  # 제목/ 내용 저장한 input()값 출력
    print(f"제목: ({command_title})")
    print(f"내용: ({command_content})")
    print(f"{i}번 글이 생성되었습니다.")

  #딕셔너리에 분류
    Article = {"num":i, "title":command_title, "content":command_content}
    Articles.append(Article)

    i += 1

  # 저장해둔 리스트 값 차례대로 출력
  elif "article_list" in command:
    if not Articles:
      print("게시글이 없다.")
      continue

    Articles.reverse()  #내용 리버스하기 - 최신순이 위로 올라가게
    print("=======================")
    print("번호   /   제목   /   내용")
    for j in Articles:
      print(f"{j.get('num')} / {j.get('title')} / {j.get('content')}")
    print("=======================")

  # 게시글 삭제 기능
  elif "article_delete" in command:
    Articles.reverse()

    for d in Articles:
      if d.get('num') == int(command[1]):
        Articles.remove(d)


  #게시글 수정 기능
  elif "article_edit" in command:
    Articles.reverse()
    Editnumber = int(command[1])
    Editcommand = (input("수정하는 것이 title인지 content인지 입력하세요.:"))

    for e in Articles:
      # 번호 입력 key와 수정하고 싶은 번호가 같은지 확인
      if e.get('num') == Editnumber:
        #리스트 내부에 있는 인덱스 값의 딕셔너리의 key 값을 수정
        Articles[Editnumber-1][f"{Editcommand}"] = input("수정할 내용 입력하세요.:")


  # 일치하지 않는 명령어일 경우 출력
  else:
    print("지원하지 않는 명령어입니다.")
```
