```python
import time


running = True
i = 4
#회원 리스트
Members = []
login = False
#게시글 클래스
class articles:
  def __init__(self,i,command_title,command_content,nickname):
    self.i = i
    self.command_title = command_title
    self.command_content = command_content
    self.nickname = nickname

  def savewrite(self):
    write = {"num":self.i, "title":self.command_title, "content":self.command_content,"owner":nickname ,"time":time.strftime('%Y-%m-%d %H:%M:%S')}
    return write

# 게시글 제한 컨트롤러
# ==============================================================================
def Postrestriction(login, command, commandBits, writes):
   # 로그인 되어 있는지 확인
  if login == False:
    print("로그인되어 있지 않아, 접근할 수 없습니다. \n[로그인]을 해주세요. [회원가입] - member join를 입력하세요. ")

  # 기입해야할 명령어 수가 부족한 경우
  elif len(command) < 3:
    print("명령어의 수가 부족합니다.")

  #만약 num을 써야할 곳에 문자열을 기입 했을 경우
  elif not commandBits[2].isdigit():
    print("해당 게시글 num이 존재하지 않습니다.")
def makeTestData():
  return[{"num":1, "title":"거제", "content":"원이","owner":"오이시" ,"time":time.strftime('2026-08-31 16:19:30')},
          {"num":2, "title":"치바", "content":"미나미","owner":"거제야호", "time":time.strftime('2026-04-14 03:19:30')},
          {"num":3, "title":"수원", "content":"리브","owner":"너도나도", "time":time.strftime('2026-05-02 11:30:30')}]
writes = makeTestData()

print("===CLI 게시판 실행===")
while running:
  # 명령어 입력
  # ============================================================================
  command = input("명령어 입력:").strip()

  # exit를 입력하면 종료
  if command == "exit":
      print("===CLI 게시판 종료===")
      break

  #로그인 (아이디, 비밀번호) 입력 요구
  # ============================================================================
  elif command == 'member login':
    #logincheck[0] = Username , logincheck[1] = password
    logincheck = input("아이디, 비밀번호를 입력해주세요.").split(",")
    found = False

    if len(command) < 2:
      print("명령어의 수가 부족합니다.")
      continue
    # 회원가입된 딕셔너리 중에 일치하는 것이 있는가?
    # 반복문을 이용해, 모두 비교
    for m in Members:

      if logincheck[0] == str(m.get("Username")) and logincheck[1] == str(m.get("Password")):
        print("로그인 되었습니다.")
        login = True
        nickname = logincheck[0]
        break

      elif found == False:
        print("회원정보와 다릅니다.")
      continue

  #로그아웃
  # ============================================================================
  elif command.startswith("member loout"):
    if login == True:
      print(f"{logincheck[0]}님 로그아웃되었습니다.")
      login = False
      logincheck = None



  #회원 가입 (아이디, 비밀번호)
  # ============================================================================
  elif command.startswith("member join"):
    usernameinput = input("아이디 입력")
    passwordinput = input("비밀번호 입력")
    passwordcheck = input("비밀번호를 다시 한번 입력해주세요.")

    idfound = False

    # 중복된 아이디가 있는지 확인 절차
    for me in Members:
      if me.get("Username") == usernameinput:
        found = True
        continue

    if idfound == True:
      print("중복된 id가 있습니다.")
      continue

    if passwordinput != passwordcheck:
      print("패스워드가 동일하지 않습니다.")
      continue

    else:
      Member = {"Username":usernameinput, "Password":passwordinput}
      Members.append(Member)
      print(f"{Member.get("Username")}님 회원가입 성공했습니다.")



  # article write를 입력하면 제목/내용 게시글 입력
  # ============================================================================
  elif command == "article write":
    # 로그인 되어 있는지 확인
    if login == False:
      print("로그인되어 있지 않아, 접근할 수 없습니다. 회원가입 - member join를 입력하세요. ")
      continue

    # 제목 입력
    print("Title write:")
    command_title = input()

    # 내용 입력
    print("Content write:")
    command_content = input()

  # 제목/ 내용 저장한 input()값 출력
    print(f"제목: ({command_title})")
    print(f"내용: ({command_content})")
    print(f"{i}번 글이 생성되었습니다.")

  #딕셔너리에 분류 (번호, 제목, 내용, 시간)
    write = articles(i,command_title,command_content,nickname)
    writes.append(write.savewrite())

    i += 1

  # 게시글 리스트 기능
  # ============================================================================
  elif command == "article list":
    today =time.strftime('%Y-%m-%d %H:%M:%S')
    # 로그인 되어 있는지 확인
    if login == False:
      print("로그인되어 있지 않아, 접근할 수 없습니다. 회원가입 - Sign up를 입력하세요. ")
      continue

    if not writes:
      print("게시글이 없다.")
      continue

    writes.reverse()  #내용 리버스하기 - 최신순이 위로 올라가게
    print("=======================")
    print("번호   /   제목   /   내용   /  작성자  /  작성일  ")

    for j in writes:
      if today.split(" ")[0] != j.get('time').split(" ")[0]:
        print(f"{j.get('num')} / {j.get('title')} / {j.get('content')} /  {j.get('owner')} /{j.get('time')}")
      else:
        print(f"{j.get('num')} / {j.get('title')} / {j.get('content')} /  {j.get('owner')} /{j.get('time')[10:]}")
    print("=======================")
    writes.reverse()


  # 게시글 삭제 기능
  # ============================================================================
  elif command.startswith("article delete"):
    commandBits = command.split(" ")

    Postrestriction(login, command, commandBits, writes)

    deletedid = int(commandBits[2])
    nicknamefound = False
    found = False

    # d에 articeles 리스트를 하나씩 넣는다.
    # num과 작성자가 같은 경우에만 삭제가 가능함
    for d in writes:
      if d.get('num') == deletedid and d.get("owner") == nickname:
        writes.remove(d)
        print(f"{commandBits[2]}을 지웠습니다.")
        nicknamefound = True
        found = True
        break

      elif d.get('num') == deletedid:
        found = True

      elif d.get("owner") == nickname:
        nicknamefound = True

    if nicknamefound == False:
      print("해당 게시글에 대한 권한이 없습니다.")

    #만약 num이 존재 하지 않는 num인 경우
    if found == False:
      print(f"{commandBits[2]}번 글은 존재하지 않습니다.")
      continue


  #게시글 수정 기능
  # ============================================================================
  elif command.startswith("article edit"):
    commandBits = command.split(" ")
    Editnumber = int(commandBits[2])
    nicknamefound = False
    found = False

    Postrestriction(login, command, commandBits, writes)
    Editcommand = (input("수정하는 것이 title인지 content인지 입력하세요.:"))

    for e in writes:
      # 번호 입력 key와 수정하고 싶은 번호가 같은지 확인
      if e.get('num') == Editnumber and e.get("owner") == nickname:
        #리스트 내부에 있는 인덱스 값의 딕셔너리의 key 값을 수정
        writes[Editnumber-1][f"{Editcommand}"] = input("수정할 내용 입력하세요.:")
        print(f"{Editnumber}번의 글에서 {Editcommand}를 수정했습니다.")
        nicknamefound = True
        found = True

      elif e.get('num') == deletedid:
        found = True

      elif e.get("owner") == nickname:
        nicknamefound = True

    if nicknamefound == False:
      print("해당 게시글에 대한 권한이 없습니다.")

    if found == False:
      print(f"{Editnumber}번 글은 존재하지 않습니다.")
      continue

  #게시글 상세보기 (번호, 제목, 내용, 날짜 )
  # ============================================================================
  elif command.startswith("article detail"):
    commandBits = command.split(" ")
    Detailnumber = int(commandBits[2])
    found = False

    Postrestriction(login, command, commandBits, writes, nickname)

    for de in writes:
      if de.get('num') == Detailnumber:
        print(f"번호: {de.get('num')} \n제목: {de.get('title')} \n내용: {de.get('content')} \n작성자:{logincheck[0]} \n시간: {de.get('time')}")
        found = True

    if found == False:
      print(f"{Detailnumber}번 글은 존재하지 않습니다.")
      continue

  # 일치하지 않는 명령어일 경우 출력
  else:
    print("지원하지 않는 명령어입니다.")
```
